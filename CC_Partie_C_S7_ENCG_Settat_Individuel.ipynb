```python
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import numpy as np

# 1. Test avec alpha = 0.1 (très peu de pénalité)
model_faible = make_pipeline(StandardScaler(), Ridge(alpha=0.1)).fit(Xr_ent, yr_ent)
coefs_faibles = model_faible[-1].coef_

# 2. Test avec alpha = 12.0 (votre code initial)
model_moyen = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
coefs_moyens = model_moyen[-1].coef_

# 3. Test avec alpha = 1000 (très forte pénalité)
model_fort = make_pipeline(StandardScaler(), Ridge(alpha=1000.0)).fit(Xr_ent, yr_ent)
coefs_forts = model_fort[-1].coef_

# Affichage de la norme L2 (somme des carrés) des coefficients pour voir l'effet
print("Somme des carrés des coefficients (norme L2) :")
print(f"Alpha = 0.1  : {np.sum(coefs_faibles**2):.2f}")
print(f"Alpha = 12.0 : {np.sum(coefs_moyens**2):.2f}")
print(f"Alpha = 1000 : {np.sum(coefs_forts**2):.2f}")

print(f"\nExemple avec le premier coefficient :")
print(f"Alpha = 0.1  : {coefs_faibles[0]:.2f}")
print(f"Alpha = 1000 : {coefs_forts[0]:.2f}")
```

Test de sensibilité et interprétation :
En testant alpha=0.1, alpha=12.0 et alpha=1000, on observe que la somme des carrés des coefficients diminue drastiquement à mesure que alpha augmente. Par exemple, en passant de alpha=0.1 à alpha=1000, le premier coefficient passe de (valeur 1) à (valeur 2, très proche de zéro).

Effet : Plus la pénalité alpha est forte, plus le modèle est contraint. Les coefficients sont "écrasés" vers zéro.
Conséquence : Le biais du modèle augmente (il devient moins précis sur les données d'entraînement), mais sa variance diminue (il généralise mieux sur les nouvelles données). Si alpha est trop grand, le modèle finit par prédire la moyenne de la cible (sous-apprentissage).

```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
from sklearn.linear_model import Lasso
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
import numpy as np

# Liste des valeurs d'alpha à tester
alphas_testes = [0.1, 2.0, 10.0]

print("--- Résultats du test de sensibilité pour Lasso ---")
for a in alphas_testes:
    # Création et entraînement du modèle
    modele_lasso = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000))
    modele_lasso.fit(Xr_ent, yr_ent)

    # Récupération des coefficients et calcul du nombre de non-nuls
    coefs = modele_lasso[-1].coef_
    nb_non_nuls = np.count_nonzero(coefs)
    nb_annules = len(coefs) - nb_non_nuls

    # Calcul de l'erreur
    mae = mean_absolute_error(yr_test, modele_lasso.predict(Xr_test))

    print(f"Alpha = {a:5} | Variables non nulles : {nb_non_nuls:2} | Variables annulées : {nb_annules:2} | MAE : {mae:.2f}")
```

Test de sensibilité et interprétation :
En exécutant le code avec trois valeurs d'alpha croissantes (0.1, 2.0 et 10.0), on observe clairement le mécanisme de Lasso :

Pour alpha = 0.1 : Le modèle conserve la majorité des variables (peu ou pas de variables annulées).
Pour alpha = 2.0 : Un certain nombre de variables ont été annulées (mises à zéro).
Pour alpha = 10.0 : Le nombre de variables annulées est maximal.
Conclusion : Plus alpha augmente, plus le modèle est pénalisé et plus il élimine de variables. Si on augmente trop alpha, le modèle finira par annuler toutes les variables et ne prédira plus que la moyenne de la cible (sous-apprentissage). L'avantage de Lasso est donc de fournir un modèle "parcimonieux" (qui utilise moins de variables), ce qui le rend plus facile à interpréter qu'un modèle Ridge.

Théorie : Augmenter le degré d'un polynôme améliore toujours (ou maintient) la performance sur l'entraînement, car le modèle gagne en flexibilité pour coller aux données.
Cependant, si le degré est trop élevé, le modèle fait du surapprentissage (overfitting) : il mémorise le bruit de l'entraînement et devient mauvais sur le test.

```python
# Vérification du surapprentissage pour la régression polynomiale (degré 2)
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

# Définition et entraînement explicite du modèle local
modele_poly = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(degree=2, include_bias=False),
    Ridge(alpha=100.0)
).fit(Xr_ent, yr_ent)

# Calcul et affichage des performances
r2_train = r2_score(yr_ent, modele_poly.predict(Xr_ent))
r2_test = r2_score(yr_test, modele_poly.predict(Xr_test))
print(f"R² entraînement : {round(r2_train, 3)}")
print(f"R² test : {round(r2_test, 3)}")
```

Interprétation : Si R² entraînement est proche de R² test (ex: 0.55 et 0.50), le degré 2 améliore réellement la généralisation. Si R² entraînement est très élevé (ex: 0.95) et le test bas (0.50), c'est du surapprentissage, le degré 2 (ou la flexibilité du modèle) est trop fort par rapport au bruit.

"Que change max_iter sur le coût et le risque de surapprentissage ?"

Sur le coût (erreur d'entraînement) : Chaque itération ajoute un nouvel arbre pour corriger les erreurs des arbres précédents. Plus max_iter est grand, plus le modèle s'ajuste finement aux données d'entraînement. Le coût (l'erreur sur le train) diminue continuellement avec l'augmentation de max_iter.
Sur le risque de surapprentissage (overfitting) : C'est le principe du compromis biais-variance. Si max_iter est trop faible, le modèle sous-apprend (biais élevé). Si max_iter est trop grand, le modèle va finir par mémoriser le bruit des données d'entraînement. En résumé : max_iter réduit le biais mais augmente la variance. Pour contrer cela, on utilise souvent un learning_rate (taux d'apprentissage) faible couplé à un max_iter élevé, et/ou une technique d'early stopping (arrêt précoce).

Pourquoi StandardScaler ? La régression logistique avec régularisation (L2 par défaut) est extrêmement sensible à l'échelle des variables. Pourquoi C=0.5 ? C'est une contrainte légèrement plus forte que la valeur par défaut (C=1.0). Cela permet de forcer le modèle à trouver une frontière de décision plus simple et de l'empêcher de se focaliser sur des valeurs aberrantes, favorisant ainsi une meilleure généralisation.

Pourquoi le f1_score ? Le F1-score est la moyenne harmonique de la précision et du rappel (Recall). C'est la métrique de choix en classification lorsque les classes sont déséquilibrées, car il pénalise les modèles qui prédisent trop souvent la classe majoritaire au détriment de la classe minoritaire. ## 26. Epsilon-greedy

**Variante C — question spécifique :** Quel effet de epsilon sur la concentration des choix ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .2
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 29. DQN — cible conceptuelle

**Variante C — question spécifique :** Pourquoi la cible est-elle égale à la récompense en état terminal ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```
